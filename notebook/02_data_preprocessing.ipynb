# Data Preprocessing

This script performs feature engineering and transformation on the cleaned healthcare dataset to prepare it for machine learning models.

```python
# Import required libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler
```

```python
# Load the previously cleaned dataset
df = pd.read_csv("../Data/health_cleaned.csv")
df.head()
```

# Dataset Overview:
- Total columns: 19 (15 original + 4 flags)
- Contains both numerical and categorical variables
- Target variable: 'Illness' (binary classification)

# One-Hot Encoding For Nominal Categorical Variable

One-Hot Encoding Strategy:

Convert nominal categorical variables (no natural order) into binary columns
drop_first=True to avoid dummy variable trap (multicollinearity)

```python
# Gender Encoding
gender_dummies = pd.get_dummies(
    df['Gender'], 
    prefix='gender', 
    dummy_na=False, 
    drop_first=True
).iloc[:, :-1].astype('int')  # Remove last column to avoid multicollinearity
```

```python
# Smoking Status Encoding
smoking_dummies = pd.get_dummies(
    df['Smoking'], 
    prefix='smoking', 
    dummy_na=False, 
    drop_first=True
).astype('int')
```

```python
# Medication Use Encoding
Medication_Use_dummies = pd.get_dummies(
    df['Medication_Use'], 
    prefix='Medication_Use', 
    dummy_na=False, 
    drop_first=True
).iloc[:, :-1].astype('int')
```

```python
# Alcohol Consumption Encoding
alcohol_dummies = pd.get_dummies(
    df['Alcohol_Consumption'], 
    prefix='Alcohol_Consumption', 
    dummy_na=False, 
    drop_first=True
).astype('int')
```

```python
# Cholesterol Level Encoding
Cholesterol_Level_dummies = pd.get_dummies(
    df['Cholesterol_Level'], 
    prefix='Cholesterol_Level', 
    dummy_na=False, 
    drop_first=True
).astype('int')
```

# Concatenat Encoded Variables


Combine all encoded dummy variables with the original dataset
Then drop the original categorical columns

```python
# Concatenate all dummy variables
df = pd.concat([
    df, 
    gender_dummies, 
    smoking_dummies, 
    Medication_Use_dummies, 
    alcohol_dummies, 
    Cholesterol_Level_dummies
], axis=1)
```

```python
# Drop original categorical columns (now encoded)
df = df.drop([
    'Gender', 
    'Smoking', 
    'Medication_Use', 
    'Alcohol_Consumption', 
    'Cholesterol_Level'
], axis=1)
```

# Ordinal Encoding For Ordinal Catergorical Variables

Ordinal Encoding Strategy:

Variables with natural order are mapped to numerical values
Preserves the inherent order in the data

```python
# Exercise Frequency Mapping (ordered from least to most frequent)
Exercise_Frequency_map = {
    "Never":1,
    "Rarely":2,
    "Often":3,
    "Daily":4
}
df['Exercise_Frequency'] = df['Exercise_Frequency'].map(Exercise_Frequency_map)

# Blood Pressure Mapping (ordered from low to high)
Blood_Pressure_map = {
    "Low":1,
    "Normal":2,
    "High":3
}
df['Blood_Pressure'] = df['Blood_Pressure'].map(Blood_Pressure_map)

# Stress Level Mapping (ordered from low to high)
Stress_Level_map = {
    "Low":1,
    "Medium":2,
    "High":3
}
df['Stress_Level'] = df['Stress_Level'].map(Stress_Level_map)
```

# Binary Encoding for Yes/No Variables

```python
# Family History encoding
df['Family_History'] = df['Family_History'].map({"No":0, "Yes":1})

# Target variable encoding (Illness)
df['Illness'] = df['Illness'].map({"No":0, "Yes":1})
```
