```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler
```

```python
df = pd.read_csv("../Data/health_cleaned.csv")
df.head()
```

```python
gender_dummies = pd.get_dummies(df['Gender'], prefix='gender', dummy_na=False).iloc[:,:-1].astype('int')
smoking_dummies = pd.get_dummies(df['Smoking'], prefix='smoking', dummy_na=False).astype('int')
Medication_Use_dummies = pd.get_dummies(df['Medication_Use'], prefix='Medication_Use', dummy_na=False).iloc[:,:-1].astype('int')
smoking_dummies = smoking_dummies.drop(columns=['smoking_Unknown'])
alcohol_dummies = pd.get_dummies(df['Alcohol_Consumption'], prefix='Alcohol_Consumption', dummy_na=False).iloc[:,:-1].astype('int')
Cholesterol_Level_dummies = pd.get_dummies(df['Cholesterol_Level'], prefix='Cholesterol_Level', dummy_na=False, drop_first=True).astype('int')
df = pd.concat([df, gender_dummies, smoking_dummies, Medication_Use_dummies, alcohol_dummies, Cholesterol_Level_dummies], axis=1)
df = df.drop(['Gender', 'Smoking', 'Medication_Use', 'Alcohol_Consumption', 'Cholesterol_Level'], axis=1)
```

```python
Exercise_Frequency_map = {
    "Never":1,
    "Rarely":2,
    "Often":3,
    "Daily":4
}
df['Exercise_Frequency'] = df['Exercise_Frequency'].map(Exercise_Frequency_map)

Blood_Pressure_map = {
    "Low":1,
    "Normal":2,
    "High":3
}
df['Blood_Pressure'] = df['Blood_Pressure'].map(Blood_Pressure_map)

Stress_Level_map = {
    "Low":1,
    "Medium":2,
    "High":3
}
df['Stress_Level'] = df['Stress_Level'].map(Stress_Level_map)
```

```python
le = LabelEncoder()
sel_col = df[['Family_History', 'Illness']]
for col in sel_col:
    df[col] = le.fit_transform(df[col])
```
